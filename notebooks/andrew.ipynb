Analysis of the Parks dataset

```python
import pandas as pd

df = pd.read_csv("../datasets/pittsburgh_parks.csv")
#print(df.head())

#we must define how we score each neighborhood in the parks submetric
#obviously, the more parks the better
#something to note is that not every listing in the dataset is a park in the traditional sense
#i.e. not a recreational park
#these parks will not be concluded in the final count

neighborhood_scores = {}

neighborhoods = df['neighborhood'].unique()

for n in neighborhoods:
    neighborhood_scores[n] = ((df['neighborhood'] == n) & (df['type'] == 'Park')).sum()
    print(f"{n} has {(df['neighborhood'] == n).sum()} parks")



```
